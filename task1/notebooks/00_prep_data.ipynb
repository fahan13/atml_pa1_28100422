```python
import sys
sys.path.append('..')
sys.path.append('../..')
sys.path.append('.')
sys.path.append('task1')

from common.seed import set_seed
from common.io_utils import save_results
from img_prep.make_subset import load_stl10, stratified_train_val_split, balanced_test_subset


set_seed(6304)

train_ds = load_stl10(split='train') 
test_ds = load_stl10(split='test')

train_idx, val_idx = stratified_train_val_split(train_ds, seed=6304)
test_subset_ids = balanced_test_subset(test_ds, seed=6304, total=500)

print(f"Train: {len(train_idx)}, Val: {len(val_idx)}, Test subset: {len(test_subset_ids)}")

save_results({
    "seed": 6304,
    "train_indices": train_idx.tolist(),
    "val_indices": val_idx.tolist(),
    "test_subset_indices": test_subset_ids
}, "../results/subset_ids.json")
```
